```python
import pandas as pd

df1 = pd.read_csv("../../data/raw/Autogenerated_BBS_points.csv", encoding="utf-8")
df2 = pd.read_csv("../../data/raw//BBS_stops_nearest_autogen_point.csv", encoding="utf-8")
```

```python
generated_points_ids = df1["rteno"].unique().tolist()
used_stops_ids = df2["rteno"].unique().tolist()

routen_ids = list(set(generated_points_ids + used_stops_ids)) 

missing_ids = [] 
used_ids = []

for rteno in generated_points_ids:
    if rteno in used_stops_ids:
        used_ids.append(rteno)
    else:
        missing_ids.append(rteno)
```
